**Decision Tree**


1  . A Decision Tree is a supervised machine‑learning model that splits
     data into branches to make predictions or decisions.- Root node_ the topmost node representing the entire dataset; it’s the starting point of the first split .- Internal nodes - intermediate splits based on features.Real‑life example: a bank uses a decision tree to decide loan approval. The root asks “Income > ₹50,000 internal nodes ask Credit score > 700 and Existing debt, and leaves decide Approve or Reject.

2  . Gini impurity- measures the probability of misclassifying a randomly
     chosen element in a subset. It ranges from 0 pure to 0.5 max impurity.
     Entropy- measures disorder using information gain; it ranges from 0 pure to 1 max impurity.
     Scikit‑learn defaults to Gini impurity because it is computationally simpler no logarithmic calculations and often yields similar results to entropy.

3  . Overfitting occurs when a tree learns the training data too well,
     capturing noise and performing poorly on unseen data.
     Detection: compare training accuracy high with testing accuracy low. A large gap indicates overfitting.

4  . Pruning reduces tree size to prevent overfitting.
     Pre-pruning- stops the tree from growing further during construction .
     Post-pruning- builds the full tree first, then removes unnecessary branches/subtrees based on validation performance.

5  . Feature importance quantifies how much each input variable   
     contributes to predictions. It helps businesses prioritize impactful factors  for strategic decisions and resource allocation.

12  . Training accuracy- the proportion of correctly classified instances
      in the training dataset.
      Testing accuracy- the proportion of correctly classified instances in the testing/validation dataset.
      Interpret-
      If training accuracy - testing accuracy, the model is overfitting   (it memorizes the training data but fails to generalize).
      If both accuracies are low, the model is underfitting (it fails to capture the underlying pattern).
